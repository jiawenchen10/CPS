```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import statsmodels.api as sm
from scipy.stats import pearsonr
import seaborn as sns
import matplotlib.patches as mpatches

# Load data
cora_sn = pd.read_csv('corrsi\\cora_gcn_vae_sn_lp_test.csv')
cora_tn = pd.read_csv('corrsi\\cora_gcn_vae_tn_lp_test.csv')

# Unique labels
unique_labels_sn = np.unique(cora_sn['measure'])
unique_labels_tn = np.unique(cora_tn['measure'])

# Generate 12 colors from the blues colormap
blues = plt.cm.Blues(np.linspace(0.1, 0.8, 12))  # Adjust the range to get a good distribution of colors
colors_sn = [mcolors.to_hex(color) for color in blues]

# Generate 12 colors from the oranges colormap
oranges = plt.cm.Oranges(np.linspace(0.1, 0.8, 12))  # Adjust the range to get a good distribution of colors
colors_tn = [mcolors.to_hex(color) for color in oranges]

# Helper function to plot each dataset with color-coded labels and regression line
def plot_scatter(ax, x_data, y_data, labels, colors, label_prefix, loc_x, loc_y):
    handles = []
    for idx, label in enumerate(np.unique(labels)):
        mask = labels == label
        scatter = ax.scatter(
            x_data[mask], y_data[mask], 
            color=colors[idx % len(colors)], alpha=0.6, s=1, label=f'{label_prefix}_{label}'
        )
        handles.append(scatter)
        
        # Fit and plot regression line
    X = sm.add_constant(x_data)
    model = sm.OLS(y_data, X).fit()
    y_pred = model.predict(X)
    ax.plot(x_data, y_pred, color=colors[0], lw=0.5)
    
    # Calculate and add overall correlation information
    corr, p_value = pearsonr(x_data, y_data)
    ax.text(loc_x, loc_y, f'Corr: {corr:.2f} \n  P-value:{p_value:.2e}', transform=ax.transAxes, fontsize=3, verticalalignment='top')
    
    return handles

# List of features to plot
features = ['sparsity', 'diff_sparsity', 'avg_degree',
       'diff_degree', 'avg_cluster', 'diff_cluster', 'avg_spl', 'diff_spl',
       'coeff_diff']


Xlabels = ['Sparsity', 'Diff Sparsity', 'Avg Degree',
       'Diff Avg Degree', 'Avg Cluster', 'Diff Avg Cluster', 'Avg Shortpath', 'Diff Avg Shortpath',
       'Diff Distribution']
def create_color_bar_legend_handle(cmap, label, width=0.3, height=0.05):
    patch = mpatches.Rectangle((0, 0), width, height, facecolor=cmap(0.5), edgecolor='black', linewidth=0.5)
    patch.set_label(label)
    return patch

# Loop through each feature and create a plot
for i, feature in enumerate(features):
    fig, axes = plt.subplots(1, 2, figsize=(2.8, 1.1), dpi=600)
    
    # Plot AUC vs feature for datasn and datatn
    handles_sn = plot_scatter(axes[0], cora_sn[feature], cora_sn['mean_auc'], cora_sn['measure'], colors_sn, 'SN', 0.5, 0.9)
    handles_tn = plot_scatter(axes[0], cora_tn[feature], cora_tn['mean_auc'], cora_tn['measure'], colors_tn, 'TN', 0.1, 0.2)
    
    # Plot AP vs feature for datasn and datatn
    plot_scatter(axes[1], cora_sn[feature], cora_sn['mean_ap'], cora_sn['measure'], colors_sn, 'SN', 0.5, 0.9)
    plot_scatter(axes[1], cora_tn[feature], cora_tn['mean_ap'], cora_tn['measure'], colors_tn, 'TN', 0.1, 0.2)
    
    # Customize subplot appearance
    for ax in axes:
        ax.grid(False)
        ax.tick_params(axis='both', labelsize=6, which='both', length=1.8, pad=1, width=0.4)
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.xaxis.set_ticks_position('bottom')
        ax.yaxis.set_ticks_position('left')
    
    # Optionally, add titles to subplotsa
    axes[0].set_xlabel(f' {Xlabels[i]}', fontsize=5)
    axes[1].set_xlabel(f'{Xlabels[i]}', fontsize=5)
    
    # Add custom color bars as legends
    handles = [
        create_color_bar_legend_handle(plt.cm.Blues, 'Sampled Nodes'),
        create_color_bar_legend_handle(plt.cm.Oranges, 'Top Nodes')
    ]
    
    # Add the custom legend to the plot
    fig.legend(handles=handles, loc='upper center', frameon=False, bbox_to_anchor=(0.95, 0.5), ncol=1, fontsize=3)
    
    # plt.tight_layout()
    fig.savefig(f'C:\\Users\\WENJIA\\Downloads\\cora_{feature}_plot.pdf', dpi=800)
    plt.show()
```
