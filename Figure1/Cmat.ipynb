```python
# Samped nodes by probability
cmat_degree_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_degree_sn.csv')
cmat_core_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_core_sn.csv')
cmat_betweenness_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_betweenness_sn.csv')
cmat_eigenvector_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_eigenvector_sn.csv')
cmat_pagerank_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_pagerank_sn.csv')
cmat_closeness_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_closeness_sn.csv')
cmat_gravitydegree_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitydegree_sn.csv')
cmat_gravitypagerank_sn  = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitypagerank_sn.csv')
cmat_gravitycore_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitycore_sn.csv')
cmat_gravitybetweenness_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitybetweenness_sn.csv')
cmat_gravityeigenvector_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravityeigenvector_sn.csv')
cmat_gravitycloseness_sn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitycloseness_sn.csv')


cmat_degree_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_degree_tn.csv')
cmat_core_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_core_tn.csv')
cmat_betweenness_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_betweenness_tn.csv')
cmat_eigenvector_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_eigenvector_tn.csv')
cmat_pagerank_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_pagerank_tn.csv')
cmat_closeness_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_closeness_tn.csv')
cmat_gravitydegree_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitydegree_tn.csv')
cmat_gravitypagerank_tn  = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitypagerank_tn.csv')
cmat_gravitycore_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitycore_tn.csv')
cmat_gravitybetweenness_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitybetweenness_tn.csv')
cmat_gravityeigenvector_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravityeigenvector_tn.csv')
cmat_gravitycloseness_tn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitycloseness_tn.csv')

cmat_degree_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_degree_rn.csv')
cmat_core_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_core_rn.csv')
cmat_betweenness_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_betweenness_rn.csv')
cmat_eigenvector_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_eigenvector_rn.csv')
cmat_pagerank_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_pagerank_rn.csv')
cmat_closeness_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_closeness_rn.csv')
cmat_gravitydegree_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitydegree_rn.csv')
cmat_gravitypagerank_rn  = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitypagerank_rn.csv')
cmat_gravitycore_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitycore_rn.csv')
cmat_gravitybetweenness_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitybetweenness_rn.csv')
cmat_gravityeigenvector_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravityeigenvector_rn.csv')
cmat_gravitycloseness_rn = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_gravitycloseness_rn.csv')

cmat_uniform = pd.read_csv('cmat/convert_accuracy/cmat_gcn_vae_degree_un.csv')



plt.style.use('nature')
from scipy.optimize import curve_fit
from scipy.interpolate import UnivariateSpline


def cubic_fit(x, a, b, c, d):
    return a * x**3 + b * x**2 + c * x + d

x = np.linspace(0,1,77)

fig,axes = plt.subplots(6,2,figsize = (2,6),dpi=600)
params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[0,0].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[0,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_degree_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, cmat_degree_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[0,0].scatter(x, cmat_degree_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_degree_tn['mean_auc'].values - cmat_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,0].fill_between(x, cmat_degree_fit - smooth_distances_mean, cmat_degree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_degree_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_degree_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x[:11], cmat_degree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,0].scatter(x[:11], cmat_degree_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_degree_rn['mean_auc'].values - cmat_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,0].fill_between(x[:11], cmat_degree_fit - smooth_distances_mean, cmat_degree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_degree_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, cmat_degree_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[0,0].scatter(x, cmat_degree_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_degree_sn['mean_auc'].values - cmat_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,0].fill_between(x, cmat_degree_fit - smooth_distances_mean, cmat_degree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[0,0].tick_params(labelsize=4)
axes[0,0].set_xticks([0,0.12,0.35,0.7,1])
axes[0,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
axes[0,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[0,0].grid (False)
axes[0,0].spines['top'].set_visible(False)
axes[0,0].spines['right'].set_visible(False)
axes[0,0].xaxis.set_ticks_position('bottom')
axes[0,0].yaxis.set_ticks_position('left')
axes[0,0].set_ylim([0.7,1])
axes[0,0].set_yticks(np.linspace(0.6, 1, 5))
axes[0,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,0].set_xticks([], minor=True)
axes[0,0].set_yticks([], minor=True)
axes[0,0].tick_params(axis='both', which='both', length=1.2)
axes[0,0].text(0.5, 0.97, 'degree', transform=axes[0,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)

params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[1,0].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[1,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_core_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, cmat_core_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[1,0].scatter(x, cmat_core_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_core_tn['mean_auc'].values - cmat_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,0].fill_between(x, cmat_core_fit - smooth_distances_mean, cmat_core_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_core_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_core_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x[:11], cmat_core_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,0].scatter(x[:11], cmat_core_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_core_rn['mean_auc'].values - cmat_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,0].fill_between(x[:11], cmat_core_fit - smooth_distances_mean, cmat_core_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_core_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, cmat_core_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[1,0].scatter(x, cmat_core_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_core_sn['mean_auc'].values - cmat_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,0].fill_between(x, cmat_core_fit - smooth_distances_mean, cmat_core_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[1,0].tick_params(labelsize=4)
axes[1,0].set_xticks([0,0.12,0.35,0.7,1])
axes[1,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[1,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[1,0].grid (False)
axes[1,0].spines['top'].set_visible(False)
axes[1,0].spines['right'].set_visible(False)
axes[1,0].xaxis.set_ticks_position('bottom')
axes[1,0].yaxis.set_ticks_position('left')
axes[1,0].set_ylim([0.7,1])
axes[1,0].set_yticks(np.linspace(0.6, 1, 5))
axes[1,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,0].set_xticks([], minor=True)
axes[1,0].set_yticks([], minor=True)
axes[1,0].tick_params(axis='both', which='both', length=1.2)
axes[1,0].text(0.5, 0.97, 'core', transform=axes[1,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[2,0].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[2,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_pagerank_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, cmat_pagerank_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[2,0].scatter(x, cmat_pagerank_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_pagerank_tn['mean_auc'].values - cmat_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,0].fill_between(x, cmat_pagerank_fit - smooth_distances_mean, cmat_pagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_pagerank_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_pagerank_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x[:11], cmat_pagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,0].scatter(x[:11], cmat_pagerank_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_pagerank_rn['mean_auc'].values - cmat_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,0].fill_between(x[:11], cmat_pagerank_fit - smooth_distances_mean, cmat_pagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_pagerank_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, cmat_pagerank_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[2,0].scatter(x, cmat_pagerank_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_pagerank_sn['mean_auc'].values - cmat_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,0].fill_between(x, cmat_pagerank_fit - smooth_distances_mean, cmat_pagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[2,0].tick_params(labelsize=4)
axes[2,0].set_xticks([0,0.12,0.35,0.7,1])
axes[2,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[2,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[2,0].grid (False)
axes[2,0].spines['top'].set_visible(False)
axes[2,0].spines['right'].set_visible(False)
axes[2,0].xaxis.set_ticks_position('bottom')
axes[2,0].yaxis.set_ticks_position('left')
axes[2,0].set_ylim([0.7,1])
axes[2,0].set_yticks(np.linspace(0.6, 1, 5))
axes[2,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,0].set_xticks([], minor=True)
axes[2,0].set_yticks([], minor=True)
axes[2,0].tick_params(axis='both', which='both', length=1.2)
axes[2,0].text(0.5, 0.97, 'pagerank', transform=axes[2,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[3,0].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[3,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_betweenness_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, cmat_betweenness_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[3,0].scatter(x, cmat_betweenness_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_betweenness_tn['mean_auc'].values - cmat_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,0].fill_between(x, cmat_betweenness_fit - smooth_distances_mean, cmat_betweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_betweenness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_betweenness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x[:11], cmat_betweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,0].scatter(x[:11], cmat_betweenness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_betweenness_rn['mean_auc'].values - cmat_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,0].fill_between(x[:11], cmat_betweenness_fit - smooth_distances_mean, cmat_betweenness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_betweenness_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, cmat_betweenness_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[3,0].scatter(x, cmat_betweenness_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_betweenness_sn['mean_auc'].values - cmat_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,0].fill_between(x, cmat_betweenness_fit - smooth_distances_mean, cmat_betweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[3,0].tick_params(labelsize=4)
axes[3,0].set_xticks([0,0.12,0.35,0.7,1])
axes[3,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[3,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[3,0].grid (False)
axes[3,0].spines['top'].set_visible(False)
axes[3,0].spines['right'].set_visible(False)
axes[3,0].xaxis.set_ticks_position('bottom')
axes[3,0].yaxis.set_ticks_position('left')
axes[3,0].set_ylim([0.7,1])
axes[3,0].set_yticks(np.linspace(0.6, 1, 5))
axes[3,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,0].set_xticks([], minor=True)
axes[3,0].set_yticks([], minor=True)
axes[3,0].tick_params(axis='both', which='both', length=1.2)
axes[3,0].text(0.5, 0.97, 'betweenness', transform=axes[3,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[4,0].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[4,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_closeness_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, cmat_closeness_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[4,0].scatter(x, cmat_closeness_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_closeness_tn['mean_auc'].values - cmat_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,0].fill_between(x, cmat_closeness_fit - smooth_distances_mean, cmat_closeness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_closeness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_closeness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x[:11], cmat_closeness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,0].scatter(x[:11], cmat_closeness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_closeness_rn['mean_auc'].values - cmat_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,0].fill_between(x[:11], cmat_closeness_fit - smooth_distances_mean, cmat_closeness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, cmat_closeness_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, cmat_closeness_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[4,0].scatter(x, cmat_closeness_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_closeness_sn['mean_auc'].values - cmat_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,0].fill_between(x, cmat_closeness_fit - smooth_distances_mean, cmat_closeness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[4,0].tick_params(labelsize=4)
axes[4,0].set_xticks([0,0.12,0.35,0.7,1])
axes[4,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[4,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[4,0].grid (False)
axes[4,0].spines['top'].set_visible(False)
axes[4,0].spines['right'].set_visible(False)
axes[4,0].xaxis.set_ticks_position('bottom')
axes[4,0].yaxis.set_ticks_position('left')
axes[4,0].set_ylim([0.7,1])
axes[4,0].set_yticks(np.linspace(0.6, 1, 5))
axes[4,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,0].set_xticks([], minor=True)
axes[4,0].set_yticks([], minor=True)
axes[4,0].tick_params(axis='both', which='both', length=1.2)
axes[4,0].text(0.5, 0.97, 'closeness', transform=axes[4,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[5,0].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[5,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_eigenvector_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, cmat_eigenvector_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[5,0].scatter(x, cmat_eigenvector_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_eigenvector_tn['mean_auc'].values - cmat_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,0].fill_between(x, cmat_eigenvector_fit - smooth_distances_mean, cmat_eigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_eigenvector_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_eigenvector_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x[:11], cmat_eigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,0].scatter(x[:11], cmat_eigenvector_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_eigenvector_rn['mean_auc'].values[:11] - cmat_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,0].fill_between(x[:11], cmat_eigenvector_fit - smooth_distances_mean, cmat_eigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_eigenvector_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, cmat_eigenvector_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[5,0].scatter(x, cmat_eigenvector_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_eigenvector_sn['mean_auc'].values - cmat_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,0].fill_between(x, cmat_eigenvector_fit - smooth_distances_mean, cmat_eigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[5,0].tick_params(labelsize=4)
axes[5,0].set_xticks([0,0.12,0.35,0.7,1])
axes[5,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[5,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[5,0].grid (False)
axes[5,0].spines['top'].set_visible(False)
axes[5,0].spines['right'].set_visible(False)
axes[5,0].xaxis.set_ticks_position('bottom')
axes[5,0].yaxis.set_ticks_position('left')
axes[5,0].set_ylim([0.7,1])
axes[5,0].set_yticks(np.linspace(0.6, 1, 5))
axes[5,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,0].set_xticks([], minor=True)
axes[5,0].set_yticks([], minor=True)
axes[5,0].tick_params(axis='both', which='both', length=1.2)
axes[5,0].text(0.5, 0.97, 'eigenvector', transform=axes[5,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[0,1].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[0,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitydegree_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, cmat_gravitydegree_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[0,1].scatter(x, cmat_gravitydegree_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitydegree_tn['mean_auc'].values - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x, cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitydegree_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:11], cmat_gravitydegree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,1].scatter(x[:11], cmat_gravitydegree_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitydegree_rn['mean_auc'].values[:11] - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x[:11], cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitydegree_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x , cmat_gravitydegree_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[0,1].scatter(x , cmat_gravitydegree_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitydegree_sn['mean_auc'].values - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x , cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[0,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_ylim([0.7,1])
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'gravitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)

params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[1,1].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[1,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitycore_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, cmat_gravitycore_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[1,1].scatter(x, cmat_gravitycore_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitycore_tn['mean_auc'].values - cmat_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,1].fill_between(x, cmat_gravitycore_fit - smooth_distances_mean, cmat_gravitycore_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitycore_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycore_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x[:11], cmat_gravitycore_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,1].scatter(x[:11], cmat_gravitycore_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitycore_rn['mean_auc'].values[:11] - cmat_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,1].fill_between(x[:11], cmat_gravitycore_fit - smooth_distances_mean, cmat_gravitycore_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitycore_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, cmat_gravitycore_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[1,1].scatter(x, cmat_gravitycore_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitycore_sn['mean_auc'].values - cmat_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,1].fill_between(x, cmat_gravitycore_fit - smooth_distances_mean, cmat_gravitycore_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[1,1].tick_params(labelsize=4)
axes[1,1].set_xticks([0,0.12,0.35,0.7,1])
axes[1,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[1,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[1,1].grid (False)
axes[1,1].spines['top'].set_visible(False)
axes[1,1].spines['right'].set_visible(False)
axes[1,1].xaxis.set_ticks_position('bottom')
axes[1,1].yaxis.set_ticks_position('left')
axes[1,1].set_ylim([0.7,1])
axes[1,1].set_yticks(np.linspace(0.6, 1, 5))
axes[1,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,1].set_xticks([], minor=True)
axes[1,1].set_yticks([], minor=True)
axes[1,1].tick_params(axis='both', which='both', length=1.2)
axes[1,1].text(0.5, 0.97, 'gravitycore', transform=axes[1,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[2,1].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[2,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitypagerank_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, cmat_gravitypagerank_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[2,1].scatter(x, cmat_gravitypagerank_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitypagerank_tn['mean_auc'].values - cmat_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,1].fill_between(x, cmat_gravitypagerank_fit - smooth_distances_mean, cmat_gravitypagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitypagerank_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitypagerank_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x[:11], cmat_gravitypagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,1].scatter(x[:11], cmat_gravitypagerank_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitypagerank_rn['mean_auc'].values[:11] - cmat_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,1].fill_between(x[:11], cmat_gravitypagerank_fit - smooth_distances_mean, cmat_gravitypagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, cmat_gravitypagerank_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, cmat_gravitypagerank_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[2,1].scatter(x, cmat_gravitypagerank_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitypagerank_sn['mean_auc'].values - cmat_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,1].fill_between(x, cmat_gravitypagerank_fit - smooth_distances_mean, cmat_gravitypagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[2,1].tick_params(labelsize=4)
axes[2,1].set_xticks([0,0.12,0.35,0.7,1])
axes[2,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[2,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[2,1].grid (False)
axes[2,1].spines['top'].set_visible(False)
axes[2,1].spines['right'].set_visible(False)
axes[2,1].xaxis.set_ticks_position('bottom')
axes[2,1].yaxis.set_ticks_position('left')
axes[2,1].set_ylim([0.7,1])
axes[2,1].set_yticks(np.linspace(0.6, 1, 5))
axes[2,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,1].set_xticks([], minor=True)
axes[2,1].set_yticks([], minor=True)
axes[2,1].tick_params(axis='both', which='both', length=1.2)
axes[2,1].text(0.5, 0.97, 'gravitypagerank', transform=axes[2,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[3,1].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[3,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitybetweenness_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitybetweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x, cmat_gravitybetweenness_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[3,1].scatter(x, cmat_gravitybetweenness_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitybetweenness_tn['mean_auc'].values - cmat_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,1].fill_between(x, cmat_gravitybetweenness_fit - smooth_distances_mean, cmat_gravitybetweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitybetweenness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitybetweenness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x[:11], cmat_gravitybetweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,1].scatter(x[:11], cmat_gravitybetweenness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitybetweenness_rn['mean_auc'].values - cmat_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,1].fill_between(x[:11], cmat_gravitybetweenness_fit - smooth_distances_mean, cmat_gravitybetweenness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitybetweenness_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitybetweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x, cmat_gravitybetweenness_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[3,1].scatter(x, cmat_gravitybetweenness_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitybetweenness_sn['mean_auc'].values - cmat_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,1].fill_between(x, cmat_gravitybetweenness_fit - smooth_distances_mean, cmat_gravitybetweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[3,1].tick_params(labelsize=4)
axes[3,1].set_xticks([0,0.12,0.35,0.7,1])
axes[3,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[3,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[3,1].grid (False)
axes[3,1].spines['top'].set_visible(False)
axes[3,1].spines['right'].set_visible(False)
axes[3,1].xaxis.set_ticks_position('bottom')
axes[3,1].yaxis.set_ticks_position('left')
axes[3,1].set_ylim([0.7,1])
axes[3,1].set_yticks(np.linspace(0.6, 1, 5))
axes[3,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,1].set_xticks([], minor=True)
axes[3,1].set_yticks([], minor=True)
axes[3,1].tick_params(axis='both', which='both', length=1.2)
axes[3,1].text(0.5, 0.97, 'gravitybetweenness', transform=axes[3,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)

params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[4,1].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[4,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitycloseness_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, cmat_gravitycloseness_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[4,1].scatter(x, cmat_gravitycloseness_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitycloseness_tn['mean_auc'].values - cmat_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,1].fill_between(x, cmat_gravitycloseness_fit - smooth_distances_mean, cmat_gravitycloseness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitycloseness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycloseness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x[:11], cmat_gravitycloseness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,1].scatter(x[:11], cmat_gravitycloseness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitycloseness_rn['mean_auc'].values - cmat_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,1].fill_between(x[:11], cmat_gravitycloseness_fit - smooth_distances_mean, cmat_gravitycloseness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitycloseness_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, cmat_gravitycloseness_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[4,1].scatter(x, cmat_gravitycloseness_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravitycloseness_sn['mean_auc'].values - cmat_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,1].fill_between(x, cmat_gravitycloseness_fit - smooth_distances_mean, cmat_gravitycloseness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[4,1].tick_params(labelsize=4)
axes[4,1].set_xticks([0,0.12,0.35,0.7,1])
axes[4,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
# axes[4,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)
axes[4,1].grid (False)
axes[4,1].spines['top'].set_visible(False)
axes[4,1].spines['right'].set_visible(False)
axes[4,1].xaxis.set_ticks_position('bottom')
axes[4,1].yaxis.set_ticks_position('left')
axes[4,1].set_ylim([0.7,1])
axes[4,1].set_yticks(np.linspace(0.6, 1, 5))
axes[4,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,1].set_xticks([], minor=True)
axes[4,1].set_yticks([], minor=True)
axes[4,1].tick_params(axis='both', which='both', length=1.2)
axes[4,1].text(0.5, 0.97, 'gravitycloseness', transform=axes[4,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[5,1].scatter(x, cmat_uniform['mean_auc'].values,color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_uniform['mean_auc'].values - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 2/4
axes[5,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravityeigenvector_tn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, cmat_gravityeigenvector_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[5,1].scatter(x, cmat_gravityeigenvector_tn['mean_auc'].values, color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravityeigenvector_tn['mean_auc'].values - cmat_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,1].fill_between(x, cmat_gravityeigenvector_fit - smooth_distances_mean, cmat_gravityeigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravityeigenvector_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravityeigenvector_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x[:11], cmat_gravityeigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,1].scatter(x[:11], cmat_gravityeigenvector_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravityeigenvector_rn['mean_auc'].values - cmat_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,1].fill_between(x[:11], cmat_gravityeigenvector_fit - smooth_distances_mean, cmat_gravityeigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravityeigenvector_sn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
cmat_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, cmat_gravityeigenvector_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[5,1].scatter(x, cmat_gravityeigenvector_sn['mean_auc'].values, color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.9)
distances = 0.4* np.abs(cmat_gravityeigenvector_sn['mean_auc'].values - cmat_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,1].fill_between(x, cmat_gravityeigenvector_fit - smooth_distances_mean, cmat_gravityeigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[5,1].grid (False)
axes[5,1].spines['top'].set_visible(False)
axes[5,1].spines['right'].set_visible(False)
axes[5,1].xaxis.set_ticks_position('bottom')
axes[5,1].yaxis.set_ticks_position('left')
axes[5,1].set_ylim([0.7,1])
axes[5,1].set_yticks(np.linspace(0.6, 1, 5))
axes[5,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,1].set_xticks([], minor=True)
axes[5,1].set_yticks([], minor=True)

axes[5,1].text(0.5, 0.97, 'gravityeigenvector', transform=axes[5,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)




plt.subplots_adjust(wspace=0.3,hspace=0.3)

for i in range(6):
    axes[i,0].set_ylabel('AUC (\%)',fontsize=3.5,labelpad=1)
    if i==5:
        axes[i,0].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
        axes[i,1].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
    if i<5:
        axes[i,0].set_ylim([0.7,1])
        axes[i,0].set_yticks(np.linspace(0.7, 1, 4))
        axes[i,0].set_yticks([0.7, 0.8, 0.9,1])



for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')
    ax.tick_params(axis='both', which='both', length=1.8,pad=1.5,width=0.35)
    ax.set_xlim([0,1])
    ax.set_xticks([0,0.21,0.433,0.65,0.866,1])
    ax.set_xticklabels(np.array([0, 5000,10000,15000,20000,23100]).astype(int))
    ax.tick_params(axis='x', rotation=45,labelsize=3.5)

plt.subplots_adjust(wspace=0.3,hspace=0.35)



fig.suptitle('Area Under the Curve (\%) on CA-Condmat',fontsize=5, y=0.9)
# plt.tight_layout()
fig.savefig('results/cmat_auc.png',dpi=350)
fig.savefig('results/cmat_auc.pdf',dpi=150)
```

```python

plt.style.use('nature')
from scipy.optimize import curve_fit
from scipy.interpolate import UnivariateSpline


def cubic_fit(x, a, b, c, d):
    return a * x**3 + b * x**2 + c * x + d

x = np.linspace(0,1,77)

fig,axes = plt.subplots(6,2,figsize = (2,6),dpi=600)

params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[0,0].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[0,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_degree[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, cmat_degree_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[0,0].scatter(x, cmat_degree[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_degree[:77, 2] - cmat_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,0].fill_between(x, cmat_degree_fit - smooth_distances_mean, cmat_degree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[0,0].tick_params(labelsize=4)

axes[0,0].set_xticks([0,0.12,0.35,0.7,1])
axes[0,0].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))

axes[0,0].grid (False)
axes[0,0].spines['top'].set_visible(False)
axes[0,0].spines['right'].set_visible(False)
axes[0,0].xaxis.set_ticks_position('bottom')
axes[0,0].yaxis.set_ticks_position('left')
axes[0,0].set_yticks(np.linspace(0.6, 1, 5))
axes[0,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,0].set_xticks([], minor=True)
axes[0,0].set_yticks([], minor=True)
axes[0,0].tick_params(axis='both', which='both', length=1.2)
axes[0,0].text(0.5, 0.97, 'Degree', transform=axes[0,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[1,0].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[1,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_core[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, cmat_core_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[1,0].scatter(x, cmat_core[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_core[:77, 2] - cmat_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[1,0].fill_between(x, cmat_core_fit - smooth_distances_mean, cmat_core_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[1,0].tick_params(labelsize=4)
axes[1,0].set_xticks([0,0.12,0.35,0.7,1])
axes[1,0].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[1,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[1,0].grid (False)
axes[1,0].spines['top'].set_visible(False)
axes[1,0].spines['right'].set_visible(False)
axes[1,0].xaxis.set_ticks_position('bottom')
axes[1,0].yaxis.set_ticks_position('left')
axes[1,0].set_yticks(np.linspace(0.6, 1, 5))
axes[1,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,0].set_xticks([], minor=True)
axes[1,0].set_yticks([], minor=True)
axes[1,0].tick_params(axis='both', which='both', length=1.2)
axes[1,0].text(0.5, 0.97, 'Core', transform=axes[1,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[2,0].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[2,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_pagerank[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, cmat_pagerank_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[2,0].scatter(x, cmat_pagerank[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_pagerank[:77, 2] - cmat_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[2,0].fill_between(x, cmat_pagerank_fit - smooth_distances_mean, cmat_pagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[2,0].tick_params(labelsize=4)
axes[2,0].set_xticks([0,0.12,0.35,0.7,1])
axes[2,0].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[2,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[2,0].grid (False)
axes[2,0].spines['top'].set_visible(False)
axes[2,0].spines['right'].set_visible(False)
axes[2,0].xaxis.set_ticks_position('bottom')
axes[2,0].yaxis.set_ticks_position('left')
axes[2,0].set_yticks(np.linspace(0.6, 1, 5))
axes[2,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,0].set_xticks([], minor=True)
axes[2,0].set_yticks([], minor=True)
axes[2,0].tick_params(axis='both', which='both', length=1.2)
axes[2,0].text(0.5, 0.97, 'Pagerank', transform=axes[2,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)

params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[3,0].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[3,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_betweenness[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, cmat_betweenness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[3,0].scatter(x, cmat_betweenness[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_betweenness[:77, 2] - cmat_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[3,0].fill_between(x, cmat_betweenness_fit - smooth_distances_mean, cmat_betweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[3,0].tick_params(labelsize=4)
axes[3,0].set_xticks([0,0.12,0.35,0.7,1])
axes[3,0].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[3,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[3,0].grid (False)
axes[3,0].spines['top'].set_visible(False)
axes[3,0].spines['right'].set_visible(False)
axes[3,0].xaxis.set_ticks_position('bottom')
axes[3,0].yaxis.set_ticks_position('left')
axes[3,0].set_yticks(np.linspace(0.6, 1, 5))
axes[3,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,0].set_xticks([], minor=True)
axes[3,0].set_yticks([], minor=True)
axes[3,0].tick_params(axis='both', which='both', length=1.2)
axes[3,0].text(0.5, 0.97, 'Betweenness', transform=axes[3,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[4,0].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[4,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_closeness[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, cmat_closeness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[4,0].scatter(x, cmat_closeness[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_closeness[:77, 2] - cmat_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[4,0].fill_between(x, cmat_closeness_fit - smooth_distances_mean, cmat_closeness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[4,0].tick_params(labelsize=4)
axes[4,0].set_xticks([0,0.12,0.35,0.7,1])
axes[4,0].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[4,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[4,0].grid (False)
axes[4,0].spines['top'].set_visible(False)
axes[4,0].spines['right'].set_visible(False)
axes[4,0].xaxis.set_ticks_position('bottom')
axes[4,0].yaxis.set_ticks_position('left')
axes[4,0].set_yticks(np.linspace(0.6, 1, 5))
axes[4,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,0].set_xticks([], minor=True)
axes[4,0].set_yticks([], minor=True)
axes[4,0].tick_params(axis='both', which='both', length=1.2)
axes[4,0].text(0.5, 0.97, 'Closeness', transform=axes[4,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[5,0].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[5,0].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_eigenvector[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, cmat_eigenvector_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[5,0].scatter(x, cmat_eigenvector[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_eigenvector[:77, 2] - cmat_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[5,0].fill_between(x, cmat_eigenvector_fit - smooth_distances_mean, cmat_eigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[5,0].tick_params(labelsize=4)
axes[5,0].set_xticks([0,0.12,0.35,0.7,1])
axes[5,0].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[5,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[5,0].grid (False)
axes[5,0].spines['top'].set_visible(False)
axes[5,0].spines['right'].set_visible(False)
axes[5,0].xaxis.set_ticks_position('bottom')
axes[5,0].yaxis.set_ticks_position('left')
axes[5,0].set_yticks(np.linspace(0.6, 1, 5))
axes[5,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,0].set_xticks([], minor=True)
axes[5,0].set_yticks([], minor=True)
axes[5,0].tick_params(axis='both', which='both', length=1.2)
axes[5,0].text(0.5, 0.97, 'Eigenvector', transform=axes[5,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)





params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)

params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[0,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[0,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:73], cmat_gravitydegree[:73, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x[:73], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:73], cmat_gravitydegree_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[0,1].scatter(x[:73], cmat_gravitydegree[:73, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitydegree[:73, 2] - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,1].fill_between(x[:73], cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'Grvaitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


axes[0,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[0,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[0,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:73], cmat_gravitydegree[:73, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x[:73], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:73], cmat_gravitydegree_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[0,1].scatter(x[:73], cmat_gravitydegree[:73, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitydegree[:73, 2] - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,1].fill_between(x[:73], cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:60], cmat_gravitydegree[:60,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x[:60], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:60], cmat_gravitydegree_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[0,1].scatter(x[:60], cmat_gravitydegree[:60, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitydegree[:60, 0] - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,1].fill_between(x[:60], cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[0,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'Grvaitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[1,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[1,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitycore[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, cmat_gravitycore_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[1,1].scatter(x, cmat_gravitycore[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitycore[:77, 2] - cmat_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) *0.5
axes[1,1].fill_between(x, cmat_gravitycore_fit - smooth_distances_mean, cmat_gravitycore_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[1,1].tick_params(labelsize=4)
axes[1,1].set_xticks([0,0.12,0.35,0.7,1])
axes[1,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[1,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[1,1].grid (False)
axes[1,1].spines['top'].set_visible(False)
axes[1,1].spines['right'].set_visible(False)
axes[1,1].xaxis.set_ticks_position('bottom')
axes[1,1].yaxis.set_ticks_position('left')
axes[1,1].set_yticks(np.linspace(0.6, 1, 5))
axes[1,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,1].set_xticks([], minor=True)
axes[1,1].set_yticks([], minor=True)
axes[1,1].tick_params(axis='both', which='both', length=1.2)
axes[1,1].text(0.5, 0.97, 'Grvaitycore', transform=axes[1,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)








params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[2,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[2,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitypagerank[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, cmat_gravitypagerank_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[2,1].scatter(x, cmat_gravitypagerank[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitypagerank[:77, 2] - cmat_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[2,1].fill_between(x, cmat_gravitypagerank_fit - smooth_distances_mean, cmat_gravitypagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[2,1].tick_params(labelsize=4)
axes[2,1].set_xticks([0,0.12,0.35,0.7,1])
axes[2,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[2,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[2,1].grid (False)
axes[2,1].spines['top'].set_visible(False)
axes[2,1].spines['right'].set_visible(False)
axes[2,1].xaxis.set_ticks_position('bottom')
axes[2,1].yaxis.set_ticks_position('left')
axes[2,1].set_yticks(np.linspace(0.6, 1, 5))
axes[2,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,1].set_xticks([], minor=True)
axes[2,1].set_yticks([], minor=True)
axes[2,1].tick_params(axis='both', which='both', length=1.2)
axes[2,1].text(0.5, 0.97, 'Gravitypagerank', transform=axes[2,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)




params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[3,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[3,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x[:60], cmat_gravitybetweenness[:60, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitybetweenness_fit = cubic_fit(x[:60], a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x[:60], cmat_gravitybetweenness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[3,1].scatter(x[:60], cmat_gravitybetweenness[:60, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitybetweenness[:60, 2] - cmat_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[3,1].fill_between(x[:60], cmat_gravitybetweenness_fit - smooth_distances_mean, cmat_gravitybetweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[3,1].tick_params(labelsize=4)
axes[3,1].set_xticks([0,0.12,0.35,0.7,1])
axes[3,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[3,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[3,1].grid (False)
axes[3,1].spines['top'].set_visible(False)
axes[3,1].spines['right'].set_visible(False)
axes[3,1].xaxis.set_ticks_position('bottom')
axes[3,1].yaxis.set_ticks_position('left')
axes[3,1].set_yticks(np.linspace(0.6, 1, 5))
axes[3,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,1].set_xticks([], minor=True)
axes[3,1].set_yticks([], minor=True)
axes[3,1].tick_params(axis='both', which='both', length=1.2)
axes[3,1].text(0.5, 0.97, 'Gravitybetweenness', transform=axes[3,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)






params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[4,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[4,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravitycloseness[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, cmat_gravitycloseness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[4,1].scatter(x, cmat_gravitycloseness[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitycloseness[:77, 2] - cmat_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[4,1].fill_between(x, cmat_gravitycloseness_fit - smooth_distances_mean, cmat_gravitycloseness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[4,1].tick_params(labelsize=4)
axes[4,1].set_xticks([0,0.12,0.35,0.7,1])
axes[4,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[4,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[4,1].grid (False)
axes[4,1].spines['top'].set_visible(False)
axes[4,1].spines['right'].set_visible(False)
axes[4,1].xaxis.set_ticks_position('bottom')
axes[4,1].yaxis.set_ticks_position('left')
axes[4,1].set_yticks(np.linspace(0.6, 1, 5))
axes[4,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,1].set_xticks([], minor=True)
axes[4,1].set_yticks([], minor=True)
axes[4,1].tick_params(axis='both', which='both', length=1.2)
axes[4,1].text(0.5, 0.97, 'Gravitycloseness', transform=axes[4,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)








params, covariance = curve_fit(cubic_fit, x, cmat_uniform[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x,cmat_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[5,1].scatter(x, cmat_uniform[:77,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_uniform[:77, 0] - cmat_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 0.5
axes[5,1].fill_between(x, cmat_uniform1-smooth_distances_mean, cmat_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, cmat_gravityeigenvector[:77, 2])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, cmat_gravityeigenvector_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[5,1].scatter(x, cmat_gravityeigenvector[:77, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravityeigenvector[:77, 2] - cmat_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[5,1].fill_between(x, cmat_gravityeigenvector_fit - smooth_distances_mean, cmat_gravityeigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[5,1].tick_params(labelsize=4)
axes[5,1].set_xticks([0,0.12,0.35,0.7,1])
axes[5,1].set_xticklabels(np.array([0, 300,770, 2000, 2700]).astype(int))
#axes[5,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[5,1].grid (False)
axes[5,1].spines['top'].set_visible(False)
axes[5,1].spines['right'].set_visible(False)
axes[5,1].xaxis.set_ticks_position('bottom')
axes[5,1].yaxis.set_ticks_position('left')
axes[5,1].set_yticks(np.linspace(0.6, 1, 5))
axes[5,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,1].set_xticks([], minor=True)
axes[5,1].set_yticks([], minor=True)
axes[5,1].tick_params(axis='both', which='both', length=1.2)
axes[5,1].text(0.5, 0.97, 'Gravityeigenvector', transform=axes[5,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)







y = np.linspace(0,1,77)
params, covariance = curve_fit(cubic_fit, x[:11], cmat_degree_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_degree_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x[:11], cmat_degree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,0].scatter(x[:11], cmat_degree_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_degree_rn['mean_auc'].values[:11] - cmat_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,0].fill_between(x[:11], cmat_degree_fit - smooth_distances_mean, cmat_degree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:11], cmat_core_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_core_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x[:11], cmat_core_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,0].scatter(x[:11], cmat_core_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_core_rn['mean_auc'].values[:11] - cmat_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[1,0].fill_between(x[:11], cmat_core_fit - smooth_distances_mean, cmat_core_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:11], cmat_pagerank_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_pagerank_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x[:11], cmat_pagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,0].scatter(x[:11], cmat_pagerank_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_pagerank_rn['mean_auc'].values[:11] - cmat_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[2,0].fill_between(x[:11], cmat_pagerank_fit - smooth_distances_mean, cmat_pagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')





params, covariance = curve_fit(cubic_fit, x[:11], cmat_betweenness_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_betweenness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x[:11], cmat_betweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,0].scatter(x[:11], cmat_betweenness_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_betweenness_rn['mean_auc'].values[:11] - cmat_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5



params, covariance = curve_fit(cubic_fit, x[:11], cmat_closeness_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_closeness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x[:11], cmat_closeness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,0].scatter(x[:11], cmat_closeness_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_closeness_rn['mean_auc'].values[:11] - cmat_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[4,0].fill_between(x[:11], cmat_closeness_fit - smooth_distances_mean, cmat_closeness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:11], cmat_eigenvector_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_eigenvector_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x[:11], cmat_eigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,0].scatter(x[:11], cmat_eigenvector_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_eigenvector_rn['mean_auc'].values[:11] - cmat_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[5,0].fill_between(x[:11], cmat_eigenvector_fit - smooth_distances_mean, cmat_eigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitydegree_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:11], cmat_gravitydegree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,1].scatter(x[:11], cmat_gravitydegree_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitydegree_rn['mean_auc'].values[:11] - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,1].fill_between(x[:11], cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitycore_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycore_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x[:11], cmat_gravitycore_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,1].scatter(x[:11], cmat_gravitycore_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances =  np.abs(cmat_gravitycore_rn['mean_auc'].values[:11] - cmat_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[1,1].fill_between(x[:11], cmat_gravitycore_fit - smooth_distances_mean, cmat_gravitycore_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')




params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitypagerank_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitypagerank_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x[:11], cmat_gravitypagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,1].scatter(x[:11], cmat_gravitypagerank_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitypagerank_rn['mean_auc'].values[:11] - cmat_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[2,1].fill_between(x[:11], cmat_gravitypagerank_fit - smooth_distances_mean, cmat_gravitypagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitybetweenness_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitybetweenness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x[:11], cmat_gravitybetweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,1].scatter(x[:11], cmat_gravitybetweenness_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitybetweenness_rn['mean_auc'].values[:11] - cmat_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[3,1].fill_between(x[:11], cmat_gravitybetweenness_fit - smooth_distances_mean, cmat_gravitybetweenness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravitycloseness_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycloseness_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x[:11], cmat_gravitycloseness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,1].scatter(x[:11], cmat_gravitycloseness_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances =  np.abs(cmat_gravitycloseness_rn['mean_auc'].values[:11] - cmat_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[4,1].fill_between(x[:11], cmat_gravitycloseness_fit - smooth_distances_mean, cmat_gravitycloseness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:11], cmat_gravityeigenvector_rn['mean_auc'].values[:11])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravityeigenvector_fit = cubic_fit(x[:11], a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x[:11], cmat_gravityeigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,1].scatter(x[:11], cmat_gravityeigenvector_rn['mean_auc'].values[:11], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravityeigenvector_rn['mean_auc'].values[:11] - cmat_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[5,1].fill_between(x[:11], cmat_gravityeigenvector_fit - smooth_distances_mean, cmat_gravityeigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, cmat_degree[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, cmat_degree_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[0,0].scatter(x, cmat_degree[:77, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_degree[:77, 0] - cmat_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,0].fill_between(x, cmat_degree_fit - smooth_distances_mean, cmat_degree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:67], cmat_core[:67,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_core_fit = cubic_fit(x[:67], a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x[:67], cmat_core_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[1,0].scatter(x[:67], cmat_core[:67, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_core[:67, 0] - cmat_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[1,0].fill_between(x[:67], cmat_core_fit - smooth_distances_mean, cmat_core_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:66], cmat_pagerank[:66,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_pagerank_fit = cubic_fit(x[:66], a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x[:66], cmat_pagerank_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[2,0].scatter(x[:66], cmat_pagerank[:66, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_pagerank[:66, 0] - cmat_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[2,0].fill_between(x[:66], cmat_pagerank_fit - smooth_distances_mean, cmat_pagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:34], cmat_betweenness[:34,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_betweenness_fit = cubic_fit(x[:34], a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x[:34], cmat_betweenness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[3,0].scatter(x[:34], cmat_betweenness[:34, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_betweenness[:34, 0] - cmat_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[3,0].fill_between(x[:34], cmat_betweenness_fit - smooth_distances_mean, cmat_betweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:66], cmat_closeness[:66,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_closeness_fit = cubic_fit(x[:66], a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x[:66], cmat_closeness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[4,0].scatter(x[:66], cmat_closeness[:66, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_closeness[:66, 0] - cmat_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[4,0].fill_between(x[:66], cmat_closeness_fit - smooth_distances_mean, cmat_closeness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:66], cmat_eigenvector[:66,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_eigenvector_fit = cubic_fit(x[:66], a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x[:66], cmat_eigenvector_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[5,0].scatter(x[:66], cmat_eigenvector[:66, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_eigenvector[:66, 0] - cmat_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[5,0].fill_between(x[:66], cmat_eigenvector_fit - smooth_distances_mean, cmat_eigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:60], cmat_gravitydegree[:60,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitydegree_fit = cubic_fit(x[:60], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:60], cmat_gravitydegree_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[0,1].scatter(x[:60], cmat_gravitydegree[:60, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitydegree[:60, 0] - cmat_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[0,1].fill_between(x[:60], cmat_gravitydegree_fit - smooth_distances_mean, cmat_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:53], cmat_gravitycore[:53,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycore_fit = cubic_fit(x[:53], a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x[:53], cmat_gravitycore_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[1,1].scatter(x[:53], cmat_gravitycore[:53, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitycore[:53, 0] - cmat_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[1,1].fill_between(x[:53], cmat_gravitycore_fit - smooth_distances_mean, cmat_gravitycore_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:60], cmat_gravitypagerank[:60,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitypagerank_fit = cubic_fit(x[:60], a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x[:60], cmat_gravitypagerank_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[2,1].scatter(x[:60], cmat_gravitypagerank[:60,0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitypagerank[:60,0] - cmat_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[2,1].fill_between(x[:60], cmat_gravitypagerank_fit - smooth_distances_mean, cmat_gravitypagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:70], cmat_gravitybetweenness[:70,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitybetweenness_fit = cubic_fit(x[:70], a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x[:70], cmat_gravitybetweenness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[3,1].scatter(x[:70], cmat_gravitybetweenness[:70, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitybetweenness[:70, 0] - cmat_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[3,1].fill_between(x[:70], cmat_gravitybetweenness_fit - smooth_distances_mean, cmat_gravitybetweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, cmat_gravitycloseness[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, cmat_gravitycloseness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[4,1].scatter(x, cmat_gravitycloseness[:77, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravitycloseness[:77, 0] - cmat_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[4,1].fill_between(x, cmat_gravitycloseness_fit - smooth_distances_mean, cmat_gravitycloseness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, cmat_gravityeigenvector[:77,0])
a_fit, b_fit, c_fit, d_fit = params
cmat_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, cmat_gravityeigenvector_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[5,1].scatter(x, cmat_gravityeigenvector[:77, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(cmat_gravityeigenvector[:77, 0] - cmat_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 0.5
axes[5,1].fill_between(x, cmat_gravityeigenvector_fit - smooth_distances_mean, cmat_gravityeigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


axes[0,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)

for i in range(6):
    axes[i,0].set_ylabel('AUC (\%)',fontsize=3.5,labelpad=1)
    if i==5:
        axes[i,0].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
        axes[i,1].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
    if i<5:
        axes[i,0].set_ylim([0.7,1])
        axes[i,0].set_yticks(np.linspace(0.7, 1, 4))
        axes[i,0].set_yticks([0.7, 0.8, 0.9,1])



for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')
    ax.tick_params(axis='both', which='both', length=1.8,pad=1.5,width=0.35)
    ax.set_xlim([0,1])
    ax.set_xticks([0,0.21,0.433,0.65,0.866,1])
    ax.set_xticklabels(np.array([0, 5000,10000,15000,20000,23100]).astype(int))
    ax.tick_params(axis='x', rotation=45,labelsize=3.5)
    ax.set_yticks(np.linspace(0.7, 1, 4))
    ax.set_yticks([0.7, 0.8, 0.9,1])

plt.subplots_adjust(wspace=0.3,hspace=0.35)



fig.suptitle('Area Under the Curve (\%) on CA-CondMat',fontsize=5, y=0.90)
fig.savefig('C:/Users/WENJIA/Downloads/papers/results/cmat_auc.png',dpi=350)
fig.savefig('C:/Users/WENJIA/Downloads/papers/results/cmat_auc.pdf',dpi=350)
```
