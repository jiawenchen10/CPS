```python

plt.style.use('nature')
from scipy.optimize import curve_fit
from scipy.interpolate import UnivariateSpline


def cubic_fit(x, a, b, c, d):
    return a * x**3 + b * x**2 + c * x + d

x = np.linspace(0,1,180)

fig,axes = plt.subplots(6,2,figsize = (2,6),dpi=600)

params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[0,0].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[0,0].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_degree[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, pubmed_degree_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[0,0].scatter(x, pubmed_degree[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_degree[:180, 2] - pubmed_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,0].fill_between(x, pubmed_degree_fit - smooth_distances_mean, pubmed_degree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[0,0].tick_params(labelsize=4)

axes[0,0].set_xticks([0,0.12,0.35,0.7,1])
axes[0,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))

axes[0,0].grid (False)
axes[0,0].spines['top'].set_visible(False)
axes[0,0].spines['right'].set_visible(False)
axes[0,0].xaxis.set_ticks_position('bottom')
axes[0,0].yaxis.set_ticks_position('left')
axes[0,0].set_yticks(np.linspace(0.6, 1, 5))
axes[0,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,0].set_xticks([], minor=True)
axes[0,0].set_yticks([], minor=True)
axes[0,0].tick_params(axis='both', which='both', length=1.2)
axes[0,0].text(0.5, 0.97, 'Degree', transform=axes[0,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[1,0].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[1,0].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_core[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, pubmed_core_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[1,0].scatter(x, pubmed_core[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_core[:180, 2] - pubmed_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,0].fill_between(x, pubmed_core_fit - smooth_distances_mean, pubmed_core_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[1,0].tick_params(labelsize=4)
axes[1,0].set_xticks([0,0.12,0.35,0.7,1])
axes[1,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[1,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[1,0].grid (False)
axes[1,0].spines['top'].set_visible(False)
axes[1,0].spines['right'].set_visible(False)
axes[1,0].xaxis.set_ticks_position('bottom')
axes[1,0].yaxis.set_ticks_position('left')
axes[1,0].set_yticks(np.linspace(0.6, 1, 5))
axes[1,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,0].set_xticks([], minor=True)
axes[1,0].set_yticks([], minor=True)
axes[1,0].tick_params(axis='both', which='both', length=1.2)
axes[1,0].text(0.5, 0.97, 'Core', transform=axes[1,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[2,0].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[2,0].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_pagerank[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, pubmed_pagerank_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[2,0].scatter(x, pubmed_pagerank[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_pagerank[:180, 2] - pubmed_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,0].fill_between(x, pubmed_pagerank_fit - smooth_distances_mean, pubmed_pagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[2,0].tick_params(labelsize=4)
axes[2,0].set_xticks([0,0.12,0.35,0.7,1])
axes[2,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[2,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[2,0].grid (False)
axes[2,0].spines['top'].set_visible(False)
axes[2,0].spines['right'].set_visible(False)
axes[2,0].xaxis.set_ticks_position('bottom')
axes[2,0].yaxis.set_ticks_position('left')
axes[2,0].set_yticks(np.linspace(0.6, 1, 5))
axes[2,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,0].set_xticks([], minor=True)
axes[2,0].set_yticks([], minor=True)
axes[2,0].tick_params(axis='both', which='both', length=1.2)
axes[2,0].text(0.5, 0.97, 'Pagerank', transform=axes[2,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)

params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[3,0].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[3,0].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_betweenness[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, pubmed_betweenness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[3,0].scatter(x, pubmed_betweenness[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_betweenness[:180, 2] - pubmed_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,0].fill_between(x, pubmed_betweenness_fit - smooth_distances_mean, pubmed_betweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[3,0].tick_params(labelsize=4)
axes[3,0].set_xticks([0,0.12,0.35,0.7,1])
axes[3,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[3,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[3,0].grid (False)
axes[3,0].spines['top'].set_visible(False)
axes[3,0].spines['right'].set_visible(False)
axes[3,0].xaxis.set_ticks_position('bottom')
axes[3,0].yaxis.set_ticks_position('left')
axes[3,0].set_yticks(np.linspace(0.6, 1, 5))
axes[3,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,0].set_xticks([], minor=True)
axes[3,0].set_yticks([], minor=True)
axes[3,0].tick_params(axis='both', which='both', length=1.2)
axes[3,0].text(0.5, 0.97, 'Betweenness', transform=axes[3,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[4,0].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[4,0].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_closeness[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, pubmed_closeness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[4,0].scatter(x, pubmed_closeness[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_closeness[:180, 2] - pubmed_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,0].fill_between(x, pubmed_closeness_fit - smooth_distances_mean, pubmed_closeness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[4,0].tick_params(labelsize=4)
axes[4,0].set_xticks([0,0.12,0.35,0.7,1])
axes[4,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[4,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[4,0].grid (False)
axes[4,0].spines['top'].set_visible(False)
axes[4,0].spines['right'].set_visible(False)
axes[4,0].xaxis.set_ticks_position('bottom')
axes[4,0].yaxis.set_ticks_position('left')
axes[4,0].set_yticks(np.linspace(0.6, 1, 5))
axes[4,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,0].set_xticks([], minor=True)
axes[4,0].set_yticks([], minor=True)
axes[4,0].tick_params(axis='both', which='both', length=1.2)
axes[4,0].text(0.5, 0.97, 'Closeness', transform=axes[4,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[5,0].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[5,0].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_eigenvector[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, pubmed_eigenvector_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[5,0].scatter(x, pubmed_eigenvector[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_eigenvector[:180, 2] - pubmed_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,0].fill_between(x, pubmed_eigenvector_fit - smooth_distances_mean, pubmed_eigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[5,0].tick_params(labelsize=4)
axes[5,0].set_xticks([0,0.12,0.35,0.7,1])
axes[5,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[5,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[5,0].grid (False)
axes[5,0].spines['top'].set_visible(False)
axes[5,0].spines['right'].set_visible(False)
axes[5,0].xaxis.set_ticks_position('bottom')
axes[5,0].yaxis.set_ticks_position('left')
axes[5,0].set_yticks(np.linspace(0.6, 1, 5))
axes[5,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,0].set_xticks([], minor=True)
axes[5,0].set_yticks([], minor=True)
axes[5,0].tick_params(axis='both', which='both', length=1.2)
axes[5,0].text(0.5, 0.97, 'Eigenvector', transform=axes[5,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)





params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)

params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[0,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[0,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitydegree[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, pubmed_gravitydegree_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[0,1].scatter(x, pubmed_gravitydegree[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitydegree[:180, 2] - pubmed_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x, pubmed_gravitydegree_fit - smooth_distances_mean, pubmed_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'Grvaitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


axes[0,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[0,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[0,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitydegree[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, pubmed_gravitydegree_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[0,1].scatter(x, pubmed_gravitydegree[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitydegree[:180, 2] - pubmed_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x, pubmed_gravitydegree_fit - smooth_distances_mean, pubmed_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitydegree[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, pubmed_gravitydegree_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[0,1].scatter(x, pubmed_gravitydegree[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitydegree[:180, 0] - pubmed_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x, pubmed_gravitydegree_fit - smooth_distances_mean, pubmed_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[0,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'Grvaitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[1,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[1,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitycore[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, pubmed_gravitycore_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[1,1].scatter(x, pubmed_gravitycore[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitycore[:180, 2] - pubmed_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) *2/4
axes[1,1].fill_between(x, pubmed_gravitycore_fit - smooth_distances_mean, pubmed_gravitycore_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[1,1].tick_params(labelsize=4)
axes[1,1].set_xticks([0,0.12,0.35,0.7,1])
axes[1,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[1,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[1,1].grid (False)
axes[1,1].spines['top'].set_visible(False)
axes[1,1].spines['right'].set_visible(False)
axes[1,1].xaxis.set_ticks_position('bottom')
axes[1,1].yaxis.set_ticks_position('left')
axes[1,1].set_yticks(np.linspace(0.6, 1, 5))
axes[1,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,1].set_xticks([], minor=True)
axes[1,1].set_yticks([], minor=True)
axes[1,1].tick_params(axis='both', which='both', length=1.2)
axes[1,1].text(0.5, 0.97, 'Grvaitycore', transform=axes[1,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)








params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[2,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[2,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitypagerank[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, pubmed_gravitypagerank_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[2,1].scatter(x, pubmed_gravitypagerank[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitypagerank[:180, 2] - pubmed_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,1].fill_between(x, pubmed_gravitypagerank_fit - smooth_distances_mean, pubmed_gravitypagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[2,1].tick_params(labelsize=4)
axes[2,1].set_xticks([0,0.12,0.35,0.7,1])
axes[2,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[2,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[2,1].grid (False)
axes[2,1].spines['top'].set_visible(False)
axes[2,1].spines['right'].set_visible(False)
axes[2,1].xaxis.set_ticks_position('bottom')
axes[2,1].yaxis.set_ticks_position('left')
axes[2,1].set_yticks(np.linspace(0.6, 1, 5))
axes[2,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,1].set_xticks([], minor=True)
axes[2,1].set_yticks([], minor=True)
axes[2,1].tick_params(axis='both', which='both', length=1.2)
axes[2,1].text(0.5, 0.97, 'Gravitypagerank', transform=axes[2,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)




params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[3,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[3,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitybetweenness[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitybetweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x, pubmed_gravitybetweenness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[3,1].scatter(x, pubmed_gravitybetweenness[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitybetweenness[:180, 2] - pubmed_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,1].fill_between(x, pubmed_gravitybetweenness_fit - smooth_distances_mean, pubmed_gravitybetweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[3,1].tick_params(labelsize=4)
axes[3,1].set_xticks([0,0.12,0.35,0.7,1])
axes[3,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[3,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[3,1].grid (False)
axes[3,1].spines['top'].set_visible(False)
axes[3,1].spines['right'].set_visible(False)
axes[3,1].xaxis.set_ticks_position('bottom')
axes[3,1].yaxis.set_ticks_position('left')
axes[3,1].set_yticks(np.linspace(0.6, 1, 5))
axes[3,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,1].set_xticks([], minor=True)
axes[3,1].set_yticks([], minor=True)
axes[3,1].tick_params(axis='both', which='both', length=1.2)
axes[3,1].text(0.5, 0.97, 'Gravitybetweenness', transform=axes[3,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)






params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[4,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[4,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravitycloseness[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, pubmed_gravitycloseness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[4,1].scatter(x, pubmed_gravitycloseness[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitycloseness[:180, 2] - pubmed_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,1].fill_between(x, pubmed_gravitycloseness_fit - smooth_distances_mean, pubmed_gravitycloseness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[4,1].tick_params(labelsize=4)
axes[4,1].set_xticks([0,0.12,0.35,0.7,1])
axes[4,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[4,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[4,1].grid (False)
axes[4,1].spines['top'].set_visible(False)
axes[4,1].spines['right'].set_visible(False)
axes[4,1].xaxis.set_ticks_position('bottom')
axes[4,1].yaxis.set_ticks_position('left')
axes[4,1].set_yticks(np.linspace(0.6, 1, 5))
axes[4,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,1].set_xticks([], minor=True)
axes[4,1].set_yticks([], minor=True)
axes[4,1].tick_params(axis='both', which='both', length=1.2)
axes[4,1].text(0.5, 0.97, 'Gravitycloseness', transform=axes[4,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)








params, covariance = curve_fit(cubic_fit, x, pubmed_uniform[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x,pubmed_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[5,1].scatter(x, pubmed_uniform[:180,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_uniform[:180, 0] - pubmed_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[5,1].fill_between(x, pubmed_uniform1-smooth_distances_mean, pubmed_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, pubmed_gravityeigenvector[:180, 2])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, pubmed_gravityeigenvector_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[5,1].scatter(x, pubmed_gravityeigenvector[:180, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravityeigenvector[:180, 2] - pubmed_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,1].fill_between(x, pubmed_gravityeigenvector_fit - smooth_distances_mean, pubmed_gravityeigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[5,1].tick_params(labelsize=4)
axes[5,1].set_xticks([0,0.12,0.35,0.7,1])
axes[5,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[5,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[5,1].grid (False)
axes[5,1].spines['top'].set_visible(False)
axes[5,1].spines['right'].set_visible(False)
axes[5,1].xaxis.set_ticks_position('bottom')
axes[5,1].yaxis.set_ticks_position('left')
axes[5,1].set_yticks(np.linspace(0.6, 1, 5))
axes[5,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,1].set_xticks([], minor=True)
axes[5,1].set_yticks([], minor=True)
axes[5,1].tick_params(axis='both', which='both', length=1.2)
axes[5,1].text(0.5, 0.97, 'Gravityeigenvector', transform=axes[5,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)







y = np.linspace(0,1,17)
params, covariance = curve_fit(cubic_fit, x[:17], pubmed_degree_rn['mean_auc'].values[:17][:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_degree_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x[:17], pubmed_degree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,0].scatter(x[:17], pubmed_degree_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_degree_rn['mean_auc'].values[:17] - pubmed_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,0].fill_between(x[:17], pubmed_degree_fit - smooth_distances_mean, pubmed_degree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x[:17], pubmed_core_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_core_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x[:17], pubmed_core_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,0].scatter(x[:17], pubmed_core_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_core_rn['mean_auc'].values[:17] - pubmed_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,0].fill_between(x[:17], pubmed_core_fit - smooth_distances_mean, pubmed_core_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:17], pubmed_pagerank_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_pagerank_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x[:17], pubmed_pagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,0].scatter(x[:17], pubmed_pagerank_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_pagerank_rn['mean_auc'].values[:17] - pubmed_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,0].fill_between(x[:17], pubmed_pagerank_fit - smooth_distances_mean, pubmed_pagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')





params, covariance = curve_fit(cubic_fit, x[:17], pubmed_betweenness_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_betweenness_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x[:17], pubmed_betweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,0].scatter(x[:17], pubmed_betweenness_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_betweenness_rn['mean_auc'].values[:17] - pubmed_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4



params, covariance = curve_fit(cubic_fit, x[:17], pubmed_closeness_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_closeness_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x[:17], pubmed_closeness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,0].scatter(x[:17], pubmed_closeness_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_closeness_rn['mean_auc'].values[:17] - pubmed_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,0].fill_between(x[:17], pubmed_closeness_fit - smooth_distances_mean, pubmed_closeness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:17], pubmed_eigenvector_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_eigenvector_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x[:17], pubmed_eigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,0].scatter(x[:17], pubmed_eigenvector_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_eigenvector_rn['mean_auc'].values[:17] - pubmed_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,0].fill_between(x[:17], pubmed_eigenvector_fit - smooth_distances_mean, pubmed_eigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:17], pubmed_gravitydegree_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitydegree_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x[:17], pubmed_gravitydegree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,1].scatter(x[:17], pubmed_gravitydegree_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitydegree_rn['mean_auc'].values[:17] - pubmed_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x[:17], pubmed_gravitydegree_fit - smooth_distances_mean, pubmed_gravitydegree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:17], pubmed_gravitycore_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitycore_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x[:17], pubmed_gravitycore_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,1].scatter(x[:17], pubmed_gravitycore_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances =  np.abs(pubmed_gravitycore_rn['mean_auc'].values[:17] - pubmed_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,1].fill_between(x[:17], pubmed_gravitycore_fit - smooth_distances_mean, pubmed_gravitycore_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')




params, covariance = curve_fit(cubic_fit, x[:17], pubmed_gravitypagerank_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitypagerank_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x[:17], pubmed_gravitypagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,1].scatter(x[:17], pubmed_gravitypagerank_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitypagerank_rn['mean_auc'].values[:17] - pubmed_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,1].fill_between(x[:17], pubmed_gravitypagerank_fit - smooth_distances_mean, pubmed_gravitypagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:17], pubmed_gravitybetweenness_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitybetweenness_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x[:17], pubmed_gravitybetweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,1].scatter(x[:17], pubmed_gravitybetweenness_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitybetweenness_rn['mean_auc'].values[:17] - pubmed_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,1].fill_between(x[:17], pubmed_gravitybetweenness_fit - smooth_distances_mean, pubmed_gravitybetweenness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, x[:17], pubmed_gravitycloseness_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitycloseness_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x[:17], pubmed_gravitycloseness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,1].scatter(x[:17], pubmed_gravitycloseness_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances =  np.abs(pubmed_gravitycloseness_rn['mean_auc'].values[:17] - pubmed_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,1].fill_between(x[:17], pubmed_gravitycloseness_fit - smooth_distances_mean, pubmed_gravitycloseness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x[:17], pubmed_gravityeigenvector_rn['mean_auc'].values[:17])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravityeigenvector_fit = cubic_fit(x[:17], a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x[:17], pubmed_gravityeigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,1].scatter(x[:17], pubmed_gravityeigenvector_rn['mean_auc'].values[:17], color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravityeigenvector_rn['mean_auc'].values[:17] - pubmed_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,1].fill_between(x[:17], pubmed_gravityeigenvector_fit - smooth_distances_mean, pubmed_gravityeigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, pubmed_degree[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, pubmed_degree_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[0,0].scatter(x, pubmed_degree[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_degree[:180, 0] - pubmed_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,0].fill_between(x, pubmed_degree_fit - smooth_distances_mean, pubmed_degree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_core[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, pubmed_core_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[1,0].scatter(x, pubmed_core[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_core[:180, 0] - pubmed_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,0].fill_between(x, pubmed_core_fit - smooth_distances_mean, pubmed_core_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_pagerank[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, pubmed_pagerank_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[2,0].scatter(x, pubmed_pagerank[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_pagerank[:180, 0] - pubmed_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,0].fill_between(x, pubmed_pagerank_fit - smooth_distances_mean, pubmed_pagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_betweenness[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, pubmed_betweenness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[3,0].scatter(x, pubmed_betweenness[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_betweenness[:180, 0] - pubmed_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,0].fill_between(x, pubmed_betweenness_fit - smooth_distances_mean, pubmed_betweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, pubmed_closeness[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, pubmed_closeness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[4,0].scatter(x, pubmed_closeness[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_closeness[:180, 0] - pubmed_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,0].fill_between(x, pubmed_closeness_fit - smooth_distances_mean, pubmed_closeness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_eigenvector[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, pubmed_eigenvector_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[5,0].scatter(x, pubmed_eigenvector[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_eigenvector[:180, 0] - pubmed_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,0].fill_between(x, pubmed_eigenvector_fit - smooth_distances_mean, pubmed_eigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, pubmed_gravitydegree[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, pubmed_gravitydegree_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[0,1].scatter(x, pubmed_gravitydegree[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitydegree[:180, 0] - pubmed_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[0,1].fill_between(x, pubmed_gravitydegree_fit - smooth_distances_mean, pubmed_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_gravitycore[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, pubmed_gravitycore_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[1,1].scatter(x, pubmed_gravitycore[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitycore[:180, 0] - pubmed_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[1,1].fill_between(x, pubmed_gravitycore_fit - smooth_distances_mean, pubmed_gravitycore_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_gravitypagerank[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, pubmed_gravitypagerank_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[2,1].scatter(x, pubmed_gravitypagerank[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitypagerank[:180, 0] - pubmed_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[2,1].fill_between(x, pubmed_gravitypagerank_fit - smooth_distances_mean, pubmed_gravitypagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_gravitybetweenness[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitybetweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x, pubmed_gravitybetweenness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[3,1].scatter(x, pubmed_gravitybetweenness[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitybetweenness[:180, 0] - pubmed_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[3,1].fill_between(x, pubmed_gravitybetweenness_fit - smooth_distances_mean, pubmed_gravitybetweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, pubmed_gravitycloseness[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, pubmed_gravitycloseness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[4,1].scatter(x, pubmed_gravitycloseness[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravitycloseness[:180, 0] - pubmed_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[4,1].fill_between(x, pubmed_gravitycloseness_fit - smooth_distances_mean, pubmed_gravitycloseness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, pubmed_gravityeigenvector[:180,0])
a_fit, b_fit, c_fit, d_fit = params
pubmed_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, pubmed_gravityeigenvector_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[5,1].scatter(x, pubmed_gravityeigenvector[:180, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(pubmed_gravityeigenvector[:180, 0] - pubmed_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 2/4
axes[5,1].fill_between(x, pubmed_gravityeigenvector_fit - smooth_distances_mean, pubmed_gravityeigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


axes[0,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)

plt.subplots_adjust(wspace=0.3,hspace=0.3)

for i in range(6):
    axes[i,0].set_ylabel('AUC (%)',fontsize=3.5,labelpad=1)
    if i==5:
        axes[i,0].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
        axes[i,1].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')
    ax.tick_params(axis='both', which='both', length=1.8,pad=1.5,width=0.4)
    ax.set_xlim([0,1])
    ax.set_xticks([0,0.08,0.167,0.335,0.49,0.65,0.81,1])
    ax.set_xticklabels(np.array([0, 1000,3000,6000,9000,12000,15000,18000]).astype(int))
    ax.set_yticks(np.linspace(0.7, 1, 4))
    ax.set_yticks([ 0.7, 0.8, 0.9,1])
    ax.tick_params(axis='x', rotation=45,labelsize=3.5)
plt.subplots_adjust(wspace=0.3,hspace=0.35)


fig.suptitle('Area Under the Curve (%) on Pubmed',fontsize=5, y=0.90)
fig.savefig('C:/Users/WENJIA/Downloads/papers/results/pubmed_auc.png',dpi=350)
fig.savefig('C:/Users/WENJIA/Downloads/papers/results/pubmed_auc.pdf',dpi=350)
```
