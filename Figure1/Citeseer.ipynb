```python

plt.style.use('nature')
from scipy.optimize import curve_fit
from scipy.interpolate import UnivariateSpline


def cubic_fit(x, a, b, c, d):
    return a * x**3 + b * x**2 + c * x + d

x = np.linspace(0,1,100)

fig,axes = plt.subplots(6,2,figsize = (2,6),dpi=600)

params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'Uniform nodes(UN*)',linewidth=0.5)
axes[0,0].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[0,0].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_degree[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, citeseer_degree_fit, color='#FFDC91FF', label='Top nodes(TN)', linewidth=0.5)
axes[0,0].scatter(x, citeseer_degree[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_degree[:100, 2] - citeseer_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,0].fill_between(x, citeseer_degree_fit - smooth_distances_mean, citeseer_degree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[0,0].tick_params(labelsize=4)

axes[0,0].set_xticks([0,0.12,0.35,0.7,1])
axes[0,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))

axes[0,0].grid (False)
axes[0,0].spines['top'].set_visible(False)
axes[0,0].spines['right'].set_visible(False)
axes[0,0].xaxis.set_ticks_position('bottom')
axes[0,0].yaxis.set_ticks_position('left')
axes[0,0].set_yticks(np.linspace(0.6, 1, 5))
axes[0,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,0].set_xticks([], minor=True)
axes[0,0].set_yticks([], minor=True)
axes[0,0].tick_params(axis='both', which='both', length=1.2)
axes[0,0].text(0.5, 0.97, 'Degree', transform=axes[0,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[1,0].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[1,0].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_core[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, citeseer_core_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[1,0].scatter(x, citeseer_core[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_core[:100, 2] - citeseer_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[1,0].fill_between(x, citeseer_core_fit - smooth_distances_mean, citeseer_core_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[1,0].tick_params(labelsize=4)
axes[1,0].set_xticks([0,0.12,0.35,0.7,1])
axes[1,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[1,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[1,0].grid (False)
axes[1,0].spines['top'].set_visible(False)
axes[1,0].spines['right'].set_visible(False)
axes[1,0].xaxis.set_ticks_position('bottom')
axes[1,0].yaxis.set_ticks_position('left')
axes[1,0].set_yticks(np.linspace(0.6, 1, 5))
axes[1,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,0].set_xticks([], minor=True)
axes[1,0].set_yticks([], minor=True)
axes[1,0].tick_params(axis='both', which='both', length=1.2)
axes[1,0].text(0.5, 0.97, 'Core', transform=axes[1,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[2,0].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[2,0].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_pagerank[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, citeseer_pagerank_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[2,0].scatter(x, citeseer_pagerank[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_pagerank[:100, 2] - citeseer_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[2,0].fill_between(x, citeseer_pagerank_fit - smooth_distances_mean, citeseer_pagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[2,0].tick_params(labelsize=4)
axes[2,0].set_xticks([0,0.12,0.35,0.7,1])
axes[2,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[2,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[2,0].grid (False)
axes[2,0].spines['top'].set_visible(False)
axes[2,0].spines['right'].set_visible(False)
axes[2,0].xaxis.set_ticks_position('bottom')
axes[2,0].yaxis.set_ticks_position('left')
axes[2,0].set_yticks(np.linspace(0.6, 1, 5))
axes[2,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,0].set_xticks([], minor=True)
axes[2,0].set_yticks([], minor=True)
axes[2,0].tick_params(axis='both', which='both', length=1.2)
axes[2,0].text(0.5, 0.97, 'Pagerank', transform=axes[2,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)

params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[3,0].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[3,0].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_betweenness[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, citeseer_betweenness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[3,0].scatter(x, citeseer_betweenness[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_betweenness[:100, 2] - citeseer_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[3,0].fill_between(x, citeseer_betweenness_fit - smooth_distances_mean, citeseer_betweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[3,0].tick_params(labelsize=4)
axes[3,0].set_xticks([0,0.12,0.35,0.7,1])
axes[3,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[3,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[3,0].grid (False)
axes[3,0].spines['top'].set_visible(False)
axes[3,0].spines['right'].set_visible(False)
axes[3,0].xaxis.set_ticks_position('bottom')
axes[3,0].yaxis.set_ticks_position('left')
axes[3,0].set_yticks(np.linspace(0.6, 1, 5))
axes[3,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,0].set_xticks([], minor=True)
axes[3,0].set_yticks([], minor=True)
axes[3,0].tick_params(axis='both', which='both', length=1.2)
axes[3,0].text(0.5, 0.97, 'Betweenness', transform=axes[3,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[4,0].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[4,0].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_closeness[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, citeseer_closeness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[4,0].scatter(x, citeseer_closeness[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_closeness[:100, 2] - citeseer_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[4,0].fill_between(x, citeseer_closeness_fit - smooth_distances_mean, citeseer_closeness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[4,0].tick_params(labelsize=4)
axes[4,0].set_xticks([0,0.12,0.35,0.7,1])
axes[4,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[4,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[4,0].grid (False)
axes[4,0].spines['top'].set_visible(False)
axes[4,0].spines['right'].set_visible(False)
axes[4,0].xaxis.set_ticks_position('bottom')
axes[4,0].yaxis.set_ticks_position('left')
axes[4,0].set_yticks(np.linspace(0.6, 1, 5))
axes[4,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,0].set_xticks([], minor=True)
axes[4,0].set_yticks([], minor=True)
axes[4,0].tick_params(axis='both', which='both', length=1.2)
axes[4,0].text(0.5, 0.97, 'Closeness', transform=axes[4,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[5,0].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[5,0].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_eigenvector[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, citeseer_eigenvector_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[5,0].scatter(x, citeseer_eigenvector[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_eigenvector[:100, 2] - citeseer_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[5,0].fill_between(x, citeseer_eigenvector_fit - smooth_distances_mean, citeseer_eigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[5,0].tick_params(labelsize=4)
axes[5,0].set_xticks([0,0.12,0.35,0.7,1])
axes[5,0].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[5,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[5,0].grid (False)
axes[5,0].spines['top'].set_visible(False)
axes[5,0].spines['right'].set_visible(False)
axes[5,0].xaxis.set_ticks_position('bottom')
axes[5,0].yaxis.set_ticks_position('left')
axes[5,0].set_yticks(np.linspace(0.6, 1, 5))
axes[5,0].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,0].set_xticks([], minor=True)
axes[5,0].set_yticks([], minor=True)
axes[5,0].tick_params(axis='both', which='both', length=1.2)
axes[5,0].text(0.5, 0.97, 'Eigenvector', transform=axes[5,0].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)





params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)

params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[0,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[0,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitydegree[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, citeseer_gravitydegree_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[0,1].scatter(x, citeseer_gravitydegree[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitydegree[:100, 2] - citeseer_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,1].fill_between(x, citeseer_gravitydegree_fit - smooth_distances_mean, citeseer_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'Grvaitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)


axes[0,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[0,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[0,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitydegree[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, citeseer_gravitydegree_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[0,1].scatter(x, citeseer_gravitydegree[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitydegree[:100, 2] - citeseer_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,1].fill_between(x, citeseer_gravitydegree_fit - smooth_distances_mean, citeseer_gravitydegree_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitydegree[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, citeseer_gravitydegree_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[0,1].scatter(x, citeseer_gravitydegree[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitydegree[:100, 0] - citeseer_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,1].fill_between(x, citeseer_gravitydegree_fit - smooth_distances_mean, citeseer_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')
axes[0,1].tick_params(labelsize=4)
axes[0,1].set_xticks([0,0.12,0.35,0.7,1])
axes[0,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[0,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[0,1].grid (False)
axes[0,1].spines['top'].set_visible(False)
axes[0,1].spines['right'].set_visible(False)
axes[0,1].xaxis.set_ticks_position('bottom')
axes[0,1].yaxis.set_ticks_position('left')
axes[0,1].set_yticks(np.linspace(0.6, 1, 5))
axes[0,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[0,1].set_xticks([], minor=True)
axes[0,1].set_yticks([], minor=True)
axes[0,1].tick_params(axis='both', which='both', length=1.2)
axes[0,1].text(0.5, 0.97, 'Grvaitydegree', transform=axes[0,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)



params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[1,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[1,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitycore[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, citeseer_gravitycore_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[1,1].scatter(x, citeseer_gravitycore[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitycore[:100, 2] - citeseer_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) *3 / 4
axes[1,1].fill_between(x, citeseer_gravitycore_fit - smooth_distances_mean, citeseer_gravitycore_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')

axes[1,1].tick_params(labelsize=4)
axes[1,1].set_xticks([0,0.12,0.35,0.7,1])
axes[1,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[1,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[1,1].grid (False)
axes[1,1].spines['top'].set_visible(False)
axes[1,1].spines['right'].set_visible(False)
axes[1,1].xaxis.set_ticks_position('bottom')
axes[1,1].yaxis.set_ticks_position('left')
axes[1,1].set_yticks(np.linspace(0.6, 1, 5))
axes[1,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[1,1].set_xticks([], minor=True)
axes[1,1].set_yticks([], minor=True)
axes[1,1].tick_params(axis='both', which='both', length=1.2)
axes[1,1].text(0.5, 0.97, 'Grvaitycore', transform=axes[1,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)








params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[2,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[2,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitypagerank[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, citeseer_gravitypagerank_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[2,1].scatter(x, citeseer_gravitypagerank[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitypagerank[:100, 2] - citeseer_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[2,1].fill_between(x, citeseer_gravitypagerank_fit - smooth_distances_mean, citeseer_gravitypagerank_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[2,1].tick_params(labelsize=4)
axes[2,1].set_xticks([0,0.12,0.35,0.7,1])
axes[2,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[2,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[2,1].grid (False)
axes[2,1].spines['top'].set_visible(False)
axes[2,1].spines['right'].set_visible(False)
axes[2,1].xaxis.set_ticks_position('bottom')
axes[2,1].yaxis.set_ticks_position('left')
axes[2,1].set_yticks(np.linspace(0.6, 1, 5))
axes[2,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[2,1].set_xticks([], minor=True)
axes[2,1].set_yticks([], minor=True)
axes[2,1].tick_params(axis='both', which='both', length=1.2)
axes[2,1].text(0.5, 0.97, 'Gravitypagerank', transform=axes[2,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)




params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[3,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[3,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitybetweenness[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitybetweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x, citeseer_gravitybetweenness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[3,1].scatter(x, citeseer_gravitybetweenness[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitybetweenness[:100, 2] - citeseer_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[3,1].fill_between(x, citeseer_gravitybetweenness_fit - smooth_distances_mean, citeseer_gravitybetweenness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[3,1].tick_params(labelsize=4)
axes[3,1].set_xticks([0,0.12,0.35,0.7,1])
axes[3,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[3,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[3,1].grid (False)
axes[3,1].spines['top'].set_visible(False)
axes[3,1].spines['right'].set_visible(False)
axes[3,1].xaxis.set_ticks_position('bottom')
axes[3,1].yaxis.set_ticks_position('left')
axes[3,1].set_yticks(np.linspace(0.6, 1, 5))
axes[3,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[3,1].set_xticks([], minor=True)
axes[3,1].set_yticks([], minor=True)
axes[3,1].tick_params(axis='both', which='both', length=1.2)
axes[3,1].text(0.5, 0.97, 'Gravitybetweenness', transform=axes[3,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)






params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[4,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[4,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravitycloseness[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, citeseer_gravitycloseness_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[4,1].scatter(x, citeseer_gravitycloseness[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitycloseness[:100, 2] - citeseer_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[4,1].fill_between(x, citeseer_gravitycloseness_fit - smooth_distances_mean, citeseer_gravitycloseness_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[4,1].tick_params(labelsize=4)
axes[4,1].set_xticks([0,0.12,0.35,0.7,1])
axes[4,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[4,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[4,1].grid (False)
axes[4,1].spines['top'].set_visible(False)
axes[4,1].spines['right'].set_visible(False)
axes[4,1].xaxis.set_ticks_position('bottom')
axes[4,1].yaxis.set_ticks_position('left')
axes[4,1].set_yticks(np.linspace(0.6, 1, 5))
axes[4,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[4,1].set_xticks([], minor=True)
axes[4,1].set_yticks([], minor=True)
axes[4,1].tick_params(axis='both', which='both', length=1.2)
axes[4,1].text(0.5, 0.97, 'Gravitycloseness', transform=axes[4,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)








params, covariance = curve_fit(cubic_fit, x, citeseer_uniform[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_uniform1 = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x,citeseer_uniform1,color ='#6F99ADFF', label = 'uniform',linewidth=0.5)
axes[5,1].scatter(x, citeseer_uniform[:100,0],color ='#6F99ADFF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_uniform[:100, 0] - citeseer_uniform1)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)])* 3 /4
axes[5,1].fill_between(x, citeseer_uniform1-smooth_distances_mean, citeseer_uniform1+smooth_distances_mean, color='#6F99ADFF', alpha=0.2,edgecolors='none')
params, covariance = curve_fit(cubic_fit, x, citeseer_gravityeigenvector[:100, 2])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, citeseer_gravityeigenvector_fit, color='#FFDC91FF', label='Top Nodes', linewidth=0.5)
axes[5,1].scatter(x, citeseer_gravityeigenvector[:100, 2], color ='#FFDC91FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravityeigenvector[:100, 2] - citeseer_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[5,1].fill_between(x, citeseer_gravityeigenvector_fit - smooth_distances_mean, citeseer_gravityeigenvector_fit + smooth_distances_mean, color='#FFDC91FF', alpha=0.2,edgecolors='none')
axes[5,1].tick_params(labelsize=4)
axes[5,1].set_xticks([0,0.12,0.35,0.7,1])
axes[5,1].set_xticklabels(np.array([0, 300,1000, 2000, 2700]).astype(int))
#axes[5,1].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='center right')
axes[5,1].grid (False)
axes[5,1].spines['top'].set_visible(False)
axes[5,1].spines['right'].set_visible(False)
axes[5,1].xaxis.set_ticks_position('bottom')
axes[5,1].yaxis.set_ticks_position('left')
axes[5,1].set_yticks(np.linspace(0.6, 1, 5))
axes[5,1].set_yticks([0.6, 0.7, 0.8, 0.9,1])
axes[5,1].set_xticks([], minor=True)
axes[5,1].set_yticks([], minor=True)
axes[5,1].tick_params(axis='both', which='both', length=1.2)
axes[5,1].text(0.5, 0.97, 'Gravityeigenvector', transform=axes[5,1].transAxes, ha='center', va='center', bbox=dict(facecolor='none', edgecolor='none', boxstyle='round', alpha=0.0), fontsize=4)







y = np.linspace(0,1,66)
params, covariance = curve_fit(cubic_fit, y, citeseer_degree_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_degree_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(y, citeseer_degree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,0].scatter(y, citeseer_degree_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_degree_rn['mean_auc'].values - citeseer_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,0].fill_between(y, citeseer_degree_fit - smooth_distances_mean, citeseer_degree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, y, citeseer_core_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_core_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(y, citeseer_core_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,0].scatter(y, citeseer_core_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_core_rn['mean_auc'].values - citeseer_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[1,0].fill_between(y, citeseer_core_fit - smooth_distances_mean, citeseer_core_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, y, citeseer_pagerank_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_pagerank_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(y, citeseer_pagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,0].scatter(y, citeseer_pagerank_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_pagerank_rn['mean_auc'].values - citeseer_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[2,0].fill_between(y, citeseer_pagerank_fit - smooth_distances_mean, citeseer_pagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')





params, covariance = curve_fit(cubic_fit, y, citeseer_betweenness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_betweenness_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(y, citeseer_betweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,0].scatter(y, citeseer_betweenness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_betweenness_rn['mean_auc'].values - citeseer_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4



params, covariance = curve_fit(cubic_fit, y, citeseer_closeness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_closeness_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(y, citeseer_closeness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,0].scatter(y, citeseer_closeness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_closeness_rn['mean_auc'].values - citeseer_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[4,0].fill_between(y, citeseer_closeness_fit - smooth_distances_mean, citeseer_closeness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, y, citeseer_eigenvector_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_eigenvector_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(y, citeseer_eigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,0].scatter(y, citeseer_eigenvector_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_eigenvector_rn['mean_auc'].values - citeseer_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[5,0].fill_between(y, citeseer_eigenvector_fit - smooth_distances_mean, citeseer_eigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, y, citeseer_gravitydegree_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitydegree_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(y, citeseer_gravitydegree_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[0,1].scatter(y, citeseer_gravitydegree_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitydegree_rn['mean_auc'].values - citeseer_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,1].fill_between(y, citeseer_gravitydegree_fit - smooth_distances_mean, citeseer_gravitydegree_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, y, citeseer_gravitycore_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitycore_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(y, citeseer_gravitycore_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[1,1].scatter(y, citeseer_gravitycore_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances =  np.abs(citeseer_gravitycore_rn['mean_auc'].values - citeseer_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[1,1].fill_between(y, citeseer_gravitycore_fit - smooth_distances_mean, citeseer_gravitycore_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')




params, covariance = curve_fit(cubic_fit, y, citeseer_gravitypagerank_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitypagerank_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(y, citeseer_gravitypagerank_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[2,1].scatter(y, citeseer_gravitypagerank_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitypagerank_rn['mean_auc'].values - citeseer_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[2,1].fill_between(y, citeseer_gravitypagerank_fit - smooth_distances_mean, citeseer_gravitypagerank_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, y, citeseer_gravitybetweenness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitybetweenness_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(y, citeseer_gravitybetweenness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[3,1].scatter(y, citeseer_gravitybetweenness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitybetweenness_rn['mean_auc'].values - citeseer_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[3,1].fill_between(y, citeseer_gravitybetweenness_fit - smooth_distances_mean, citeseer_gravitybetweenness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')



params, covariance = curve_fit(cubic_fit, y, citeseer_gravitycloseness_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitycloseness_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(y, citeseer_gravitycloseness_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[4,1].scatter(y, citeseer_gravitycloseness_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances =  np.abs(citeseer_gravitycloseness_rn['mean_auc'].values - citeseer_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[4,1].fill_between(y, citeseer_gravitycloseness_fit - smooth_distances_mean, citeseer_gravitycloseness_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, y, citeseer_gravityeigenvector_rn['mean_auc'].values)
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravityeigenvector_fit = cubic_fit(y, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(y, citeseer_gravityeigenvector_fit, color='#A392BA', label='Rejected nodes(RN)', linewidth=0.5)
axes[5,1].scatter(y, citeseer_gravityeigenvector_rn['mean_auc'].values, color ='#A392BA',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravityeigenvector_rn['mean_auc'].values - citeseer_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[5,1].fill_between(y, citeseer_gravityeigenvector_fit - smooth_distances_mean, citeseer_gravityeigenvector_fit + smooth_distances_mean, color='#A392BA', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, citeseer_degree[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_degree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,0].plot(x, citeseer_degree_fit, color='#BC3C29FF', label='Sampled nodes(SN)', linewidth=0.5)
axes[0,0].scatter(x, citeseer_degree[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_degree[:100, 0] - citeseer_degree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,0].fill_between(x, citeseer_degree_fit - smooth_distances_mean, citeseer_degree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_core[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_core_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,0].plot(x, citeseer_core_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[1,0].scatter(x, citeseer_core[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_core[:100, 0] - citeseer_core_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[1,0].fill_between(x, citeseer_core_fit - smooth_distances_mean, citeseer_core_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_pagerank[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_pagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,0].plot(x, citeseer_pagerank_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[2,0].scatter(x, citeseer_pagerank[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_pagerank[:100, 0] - citeseer_pagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[2,0].fill_between(x, citeseer_pagerank_fit - smooth_distances_mean, citeseer_pagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_betweenness[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_betweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,0].plot(x, citeseer_betweenness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[3,0].scatter(x, citeseer_betweenness[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_betweenness[:100, 0] - citeseer_betweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[3,0].fill_between(x, citeseer_betweenness_fit - smooth_distances_mean, citeseer_betweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, citeseer_closeness[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_closeness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,0].plot(x, citeseer_closeness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[4,0].scatter(x, citeseer_closeness[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_closeness[:100, 0] - citeseer_closeness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[4,0].fill_between(x, citeseer_closeness_fit - smooth_distances_mean, citeseer_closeness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_eigenvector[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_eigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,0].plot(x, citeseer_eigenvector_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[5,0].scatter(x, citeseer_eigenvector[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_eigenvector[:100, 0] - citeseer_eigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[5,0].fill_between(x, citeseer_eigenvector_fit - smooth_distances_mean, citeseer_eigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, citeseer_gravitydegree[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitydegree_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[0,1].plot(x, citeseer_gravitydegree_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[0,1].scatter(x, citeseer_gravitydegree[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitydegree[:100, 0] - citeseer_gravitydegree_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[0,1].fill_between(x, citeseer_gravitydegree_fit - smooth_distances_mean, citeseer_gravitydegree_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_gravitycore[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitycore_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[1,1].plot(x, citeseer_gravitycore_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[1,1].scatter(x, citeseer_gravitycore[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitycore[:100, 0] - citeseer_gravitycore_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[1,1].fill_between(x, citeseer_gravitycore_fit - smooth_distances_mean, citeseer_gravitycore_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_gravitypagerank[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitypagerank_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[2,1].plot(x, citeseer_gravitypagerank_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[2,1].scatter(x, citeseer_gravitypagerank[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitypagerank[:100, 0] - citeseer_gravitypagerank_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[2,1].fill_between(x, citeseer_gravitypagerank_fit - smooth_distances_mean, citeseer_gravitypagerank_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_gravitybetweenness[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitybetweenness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[3,1].plot(x, citeseer_gravitybetweenness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[3,1].scatter(x, citeseer_gravitybetweenness[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitybetweenness[:100, 0] - citeseer_gravitybetweenness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[3,1].fill_between(x, citeseer_gravitybetweenness_fit - smooth_distances_mean, citeseer_gravitybetweenness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


params, covariance = curve_fit(cubic_fit, x, citeseer_gravitycloseness[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravitycloseness_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[4,1].plot(x, citeseer_gravitycloseness_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[4,1].scatter(x, citeseer_gravitycloseness[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravitycloseness[:100, 0] - citeseer_gravitycloseness_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[4,1].fill_between(x, citeseer_gravitycloseness_fit - smooth_distances_mean, citeseer_gravitycloseness_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')

params, covariance = curve_fit(cubic_fit, x, citeseer_gravityeigenvector[:100,0])
a_fit, b_fit, c_fit, d_fit = params
citeseer_gravityeigenvector_fit = cubic_fit(x, a_fit, b_fit, c_fit, d_fit)
axes[5,1].plot(x, citeseer_gravityeigenvector_fit, color='#BC3C29FF', label='Sampled Nodes', linewidth=0.5)
axes[5,1].scatter(x, citeseer_gravityeigenvector[:100, 0], color ='#BC3C29FF',edgecolors='none', s=0.4, alpha=0.6)
distances = np.abs(citeseer_gravityeigenvector[:100, 0] - citeseer_gravityeigenvector_fit)
spline = UnivariateSpline(np.arange(len(distances)), distances.flatten(), s=0)
smooth_distances = spline(np.arange(len(distances)))
smooth_distances_mean = np.mean([np.min(smooth_distances), np.max(smooth_distances)]) * 3 / 4
axes[5,1].fill_between(x, citeseer_gravityeigenvector_fit - smooth_distances_mean, citeseer_gravityeigenvector_fit + smooth_distances_mean, color='#BC3C29FF', alpha=0.2,edgecolors='none')


axes[0,0].legend(fontsize=3,edgecolor='black',framealpha=0.1,loc='best',frameon=False)

plt.subplots_adjust(wspace=0.3,hspace=0.3)

for i in range(6):
    axes[i,0].set_ylabel('AUC (%)',fontsize=3.5,labelpad=1)
    if i==5:
        axes[i,0].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)
        axes[i,1].set_xlabel('Nodes Number',fontsize=3.5,labelpad=1)

for ax in axes.flat:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.xaxis.set_ticks_position('bottom')
    ax.yaxis.set_ticks_position('left')
    ax.tick_params(axis='both', which='both', length=1.8,pad=1.5,width=0.4)
    ax.set_xlim([0,1])
    ax.set_xticks([0,0.08,0.227,0.35,0.53,0.7,0.85,1])
    ax.set_xticklabels(np.array([0, 300,750,1200,1800,2400,2850,3300]).astype(int))
    ax.set_yticks(np.linspace(0.6, 1, 5))
    ax.set_yticks([0.6, 0.7, 0.8, 0.9,1])
    ax.tick_params(axis='x', rotation=45,labelsize=3.5)

plt.subplots_adjust(wspace=0.3,hspace=0.35)


fig.suptitle('Area Under the Curve (%) on Citeseer',fontsize=5, y=0.90)
fig.savefig('C:/Users/WENJIA/Downloads/papers/results/citeseer_auc.png',dpi=350)
fig.savefig('C:/Users/WENJIA/Downloads/papers/results/citeseer_auc.pdf',dpi=350)
```
